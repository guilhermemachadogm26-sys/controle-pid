```python
%%writefile cstr_model.py
"""
Modelo de CSTR nao-isotermico (reacao exotermica irreversivel A -> B).
Parametros classicos de benchmark de controle de processos (mesma
familia usada em livros-texto de controle de processos quimicos),
que exibe nao-linearidade forte e multiplicidade de estados
estacionarios - cenario interessante para comparar controladores
lineares (PID) com agentes que aprendem politicas nao-lineares.
"""
import numpy as np

class CSTR:
    def __init__(self):
        # Parametros do processo (unidades: min, L, mol, K, J)
        self.q = 100.0        # vazao volumetrica [L/min]
        self.V = 100.0        # volume do reator [L]
        self.rho = 1000.0     # densidade [g/L]
        self.Cp = 0.239       # capacidade termica [J/(g.K)]
        self.dHr = -5.0e4     # entalpia de reacao [J/mol]  (exotermica)
        self.EoverR = 8750.0  # E/R [K]
        self.k0 = 7.2e10      # fator pre-exponencial [1/min]
        self.UA = 5.0e4       # coef. global x area de troca [J/(min.K)]
        self.Caf = 1.0        # concentracao de alimentacao [mol/L]
        self.Tf = 350.0       # temperatura de alimentacao [K]

        # Faixa fisica da variavel manipulada (temperatura do fluido
        # de resfriamento na camisa)
        self.Tc_min = 280.0
        self.Tc_max = 350.0

    def k(self, T):
        return self.k0 * np.exp(-self.EoverR / T)

    def derivs(self, state, Tc):
        Ca, T = state
        Ca = max(Ca, 1e-9)  # positividade (concentracao nao pode ser negativa)
        rate = self.k(T) * Ca
        dCadt = (self.q / self.V) * (self.Caf - Ca) - rate
        dTdt = ((self.q / self.V) * (self.Tf - T)
                + (-self.dHr) / (self.rho * self.Cp) * rate
                + (self.UA) / (self.V * self.rho * self.Cp) * (Tc - T))
        return np.array([dCadt, dTdt])

    def step(self, state, Tc, dt, n_sub=10):
        """Integracao por Runge-Kutta 4 com sub-passos para estabilidade."""
        h = dt / n_sub
        s = np.array(state, dtype=float)
        for _ in range(n_sub):
            k1 = self.derivs(s, Tc)
            k2 = self.derivs(s + h/2*k1, Tc)
            k3 = self.derivs(s + h/2*k2, Tc)
            k4 = self.derivs(s + h*k3, Tc)
            s = s + (h/6.0)*(k1 + 2*k2 + 2*k3 + k4)
            s[0] = max(s[0], 1e-9)
        return s

    def steady_state_near(self, Tc, Ca0=0.5, T0=350.0, dt=0.01, n_steps=20000):
        """Roda o modelo ate convergir, para achar um estado estacionario
        proximo ao chute inicial dado (util pela multiplicidade de
        estados estacionarios deste sistema)."""
        s = np.array([Ca0, T0], dtype=float)
        for _ in range(n_steps):
            s = self.step(s, Tc, dt, n_sub=1)
        return s
```

```python
%%writefile cstr_env.py
"""
Ambiente de controle do CSTR no estilo Gymnasium.
Observacao: [T_normalizada, Ca_normalizada, setpoint_normalizado, erro_normalizado]
Acao: variacao normalizada de Tc, em [-1, 1], mapeada para a faixa fisica de Tc.
Recompensa: penaliza erro de rastreamento e movimento excessivo do atuador.
"""
import numpy as np
import gymnasium as gym
from gymnasium import spaces
from cstr_model import CSTR

class CSTREnv(gym.Env):
    metadata = {"render_modes": []}

    def __init__(self, dt=0.05, episode_minutes=15.0, randomize_setpoint=True, seed=None):
        super().__init__()
        self.plant = CSTR()
        self.dt = dt
        self.n_steps = int(episode_minutes / dt)
        self.randomize_setpoint = randomize_setpoint
        self.rng = np.random.default_rng(seed)

        # Normalizacao (faixas tipicas de operacao deste processo)
        self.T_lo, self.T_hi = 300.0, 450.0
        self.Ca_lo, self.Ca_hi = 0.0, 1.0

        self.observation_space = spaces.Box(low=-2.0, high=2.0, shape=(4,), dtype=np.float32)
        self.action_space = spaces.Box(low=-1.0, high=1.0, shape=(1,), dtype=np.float32)

        self.Tc_center = (self.plant.Tc_min + self.plant.Tc_max) / 2.0
        self.Tc_half_range = (self.plant.Tc_max - self.plant.Tc_min) / 2.0
        self.max_dTc_per_step = 5.0  # limite de taxa de variacao do atuador [K/passo]

    def _norm_T(self, T):
        return 2 * (T - self.T_lo) / (self.T_hi - self.T_lo) - 1

    def _norm_Ca(self, Ca):
        return 2 * (Ca - self.Ca_lo) / (self.Ca_hi - self.Ca_lo) - 1

    def _obs(self):
        Ca, T = self.state
        e = (self.setpoint - T) / 50.0  # erro normalizado (escala ~K)
        return np.array([self._norm_T(T), self._norm_Ca(Ca),
                          self._norm_T(self.setpoint), e], dtype=np.float32)

    def reset(self, *, seed=None, options=None, Ca0=None, T0=None, setpoint=None):
        super().reset(seed=seed)
        if seed is not None:
            self.rng = np.random.default_rng(seed)
        Ca0 = 0.5 if Ca0 is None else Ca0
        T0 = 350.0 if T0 is None else T0
        self.state = np.array([Ca0, T0], dtype=float)
        self.Tc_prev = self.Tc_center
        if setpoint is not None:
            self.setpoint = setpoint
        elif self.randomize_setpoint:
            self.setpoint = self.rng.uniform(345.0, 375.0)
        else:
            self.setpoint = 360.0
        self.t = 0
        return self._obs(), {}

    def step(self, action):
        a = float(np.clip(action[0], -1.0, 1.0))
        Tc_target = self.Tc_center + a * self.Tc_half_range
        # limita taxa de variacao (atuador fisico realista)
        dTc = np.clip(Tc_target - self.Tc_prev, -self.max_dTc_per_step, self.max_dTc_per_step)
        Tc = self.Tc_prev + dTc
        Tc = np.clip(Tc, self.plant.Tc_min, self.plant.Tc_max)
        self.Tc_prev = Tc

        self.state = self.plant.step(self.state, Tc, self.dt)
        Ca, T = self.state
        e = self.setpoint - T

        reward = -(e ** 2) / 100.0 - 0.01 * (dTc ** 2)
        self.t += 1
        terminated = False
        truncated = self.t >= self.n_steps
        info = {"T": T, "Ca": Ca, "Tc": Tc, "setpoint": self.setpoint, "error": e}
        return self._obs(), reward, terminated, truncated, info
```

```python
%%writefile pid_controller.py
"""PID discreto classico, com anti-windup por clamping e filtro na derivada."""
import numpy as np

class PID:
    def __init__(self, Kp, Ki, Kd, dt, u_min=-1.0, u_max=1.0, d_filter=0.1):
        self.Kp, self.Ki, self.Kd = Kp, Ki, Kd
        self.dt = dt
        self.u_min, self.u_max = u_min, u_max
        self.d_filter = d_filter
        self.reset()

    def reset(self):
        self.integral = 0.0
        self.prev_error = None
        self.prev_deriv = 0.0

    def compute(self, setpoint, measurement):
        error = setpoint - measurement
        deriv = 0.0 if self.prev_error is None else (error - self.prev_error) / self.dt
        deriv = self.d_filter * deriv + (1 - self.d_filter) * self.prev_deriv
        self.prev_deriv = deriv

        u_unclamped = self.Kp * error + self.Ki * self.integral + self.Kd * deriv
        u = np.clip(u_unclamped, self.u_min, self.u_max)

        # anti-windup: só integra se não estiver saturado (ou se ajudar a sair da saturação)
        if u == u_unclamped or (u == self.u_max and error < 0) or (u == self.u_min and error > 0):
            self.integral += error * self.dt

        self.prev_error = error
        return u
```

```python
%%writefile policy.py
"""Politica neural pequena (MLP), com forward pass em numpy puro.
Isso evita depender de frameworks pesados (PyTorch/TensorFlow) para
este exemplo didatico - o agente eh treinado com um metodo de
otimizacao de caixa-preta (Cross-Entropy Method), que nao precisa
de gradientes/backpropagation."""
import numpy as np

class MLPPolicy:
    def __init__(self, in_dim=4, hidden=8, out_dim=1):
        self.in_dim, self.hidden, self.out_dim = in_dim, hidden, out_dim
        self.n_params = in_dim*hidden + hidden + hidden*out_dim + out_dim

    def unpack(self, theta):
        i = 0
        W1 = theta[i:i+self.in_dim*self.hidden].reshape(self.in_dim, self.hidden); i += self.in_dim*self.hidden
        b1 = theta[i:i+self.hidden]; i += self.hidden
        W2 = theta[i:i+self.hidden*self.out_dim].reshape(self.hidden, self.out_dim); i += self.hidden*self.out_dim
        b2 = theta[i:i+self.out_dim]; i += self.out_dim
        return W1, b1, W2, b2

    def act(self, obs, theta):
        W1, b1, W2, b2 = self.unpack(theta)
        h = np.tanh(obs @ W1 + b1)
        out = np.tanh(h @ W2 + b2)
        return out
```

```python
%%writefile train_cem.py
"""
Treina a politica MLPPolicy usando o Cross-Entropy Method (CEM):
um metodo de otimizacao evolutiva simples e robusto, adequado para
problemas de controle continuo de baixa dimensao como este.

Loop do CEM:
  1. Amostra N vetores de parametros theta ~ N(mu, sigma)
  2. Avalia cada um rodando um episodio completo no ambiente (recompensa acumulada)
  3. Seleciona os K melhores (elite)
  4. Atualiza mu e sigma a partir da elite
  5. Repete por G geracoes
"""
import numpy as np
from cstr_env import CSTREnv
from policy import MLPPolicy

def rollout(env, policy, theta, setpoint=None, seed=None, Ca0=None, T0=None):
    obs, _ = env.reset(seed=seed, setpoint=setpoint, Ca0=Ca0, T0=T0)
    total_reward = 0.0
    for _ in range(env.n_steps):
        action = policy.act(obs, theta)
        obs, r, term, trunc, info = env.step(action)
        total_reward += r
        if term or trunc:
            break
    return total_reward

def train(generations=30, pop_size=24, elite_frac=0.25, seed=0,
          setpoints_per_eval=(350, 365, 375), checkpoint_path="checkpoint.npz",
          resume=True):
    rng = np.random.default_rng(seed)
    env = CSTREnv(randomize_setpoint=False)
    policy = MLPPolicy()
    n_params = policy.n_params
    n_elite = max(2, int(pop_size * elite_frac))

    start_gen = 0
    mu = np.zeros(n_params)
    sigma = np.ones(n_params) * 0.5
    history = []

    if resume:
        try:
            data = np.load(checkpoint_path, allow_pickle=True)
            mu = data["mu"]
            sigma = data["sigma"]
            start_gen = int(data["gen"]) + 1
            history = list(data["history"])
            print(f"Retomando do checkpoint na geracao {start_gen}")
        except FileNotFoundError:
            pass

    for gen in range(start_gen, generations):
        thetas = mu + sigma * rng.standard_normal((pop_size, n_params))
        fitness = np.zeros(pop_size)
        for i in range(pop_size):
            total = 0.0
            for sp in setpoints_per_eval:
                total += rollout(env, policy, thetas[i], setpoint=sp, seed=gen*1000+i)
            fitness[i] = total / len(setpoints_per_eval)

        elite_idx = np.argsort(fitness)[-n_elite:]
        elite = thetas[elite_idx]
        mu = elite.mean(axis=0)
        sigma = elite.std(axis=0) + 0.02  # piso de exploracao para nao colapsar cedo demais

        best = fitness[elite_idx[-1]]
        mean_fit = fitness.mean()
        history.append((gen, best, mean_fit))
        print(f"Geracao {gen+1:3d}/{generations} | melhor recompensa={best:9.2f} | media={mean_fit:9.2f}", flush=True)

        np.savez(checkpoint_path, mu=mu, sigma=sigma, gen=gen,
                 history=np.array(history, dtype=float))

    return mu, policy, history

if __name__ == "__main__":
    best_theta, policy, history = train()
    np.save("trained_policy.npy", best_theta)
    print("Politica treinada salva em trained_policy.npy")
```

```python
%%writefile compare.py
"""
Compara o agente treinado (CEM) com o PID sintonizado no mesmo CSTR,
para varios setpoints de teste. Gera:
  - grafico de resposta no tempo (temperatura e acao de controle)
  - tabela de metricas: ISE, IAE, overshoot, tempo de acomodacao
"""
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

from cstr_env import CSTREnv
from pid_controller import PID
from policy import MLPPolicy

PID_GAINS = dict(Kp=0.0006, Ki=0.2, Kd=0.16)
TEST_SETPOINTS = [345.0, 360.0, 375.0]  # inclui um ponto nao visto no treino (360)

def run_pid(setpoint, T0=350.0, Ca0=0.5):
    env = CSTREnv(randomize_setpoint=False)
    pid = PID(**PID_GAINS, dt=env.dt)
    obs, _ = env.reset(setpoint=setpoint, T0=T0, Ca0=Ca0)
    pid.reset()
    log = {"t": [], "T": [], "Tc": [], "error": []}
    for k in range(env.n_steps):
        T = env.state[1]
        u = pid.compute(setpoint, T)
        obs, r, term, trunc, info = env.step([u])
        log["t"].append((k+1)*env.dt)
        log["T"].append(info["T"])
        log["Tc"].append(info["Tc"])
        log["error"].append(info["error"])
    return {k: np.array(v) for k, v in log.items()}

def run_agent(setpoint, theta, T0=350.0, Ca0=0.5):
    env = CSTREnv(randomize_setpoint=False)
    policy = MLPPolicy()
    obs, _ = env.reset(setpoint=setpoint, T0=T0, Ca0=Ca0)
    log = {"t": [], "T": [], "Tc": [], "error": []}
    for k in range(env.n_steps):
        action = policy.act(obs, theta)
        obs, r, term, trunc, info = env.step(action)
        log["t"].append((k+1)*env.dt)
        log["T"].append(info["T"])
        log["Tc"].append(info["Tc"])
        log["error"].append(info["error"])
    return {k: np.array(v) for k, v in log.items()}

def metrics(log, setpoint, dt, tol=0.02):
    e = log["error"]
    T = log["T"]
    t = log["t"]
    ise = np.sum(e**2) * dt
    iae = np.sum(np.abs(e)) * dt
    overshoot = max(0.0, (T.max() - setpoint) if setpoint >= T[0] else (setpoint - T.min()))
    band = tol * abs(setpoint)
    within = np.abs(T - setpoint) <= band
    settle_idx = None
    for i in range(len(within)):
        if np.all(within[i:]):
            settle_idx = i
            break
    settling_time = t[settle_idx] if settle_idx is not None else float("nan")
    return dict(ISE=ise, IAE=iae, overshoot_K=overshoot, settling_time_min=settling_time)

def main():
    theta = np.load("trained_policy.npy")
    env_dt = CSTREnv().dt

    fig, axes = plt.subplots(2, len(TEST_SETPOINTS), figsize=(5*len(TEST_SETPOINTS), 7), sharex=True)
    rows = []
    for j, sp in enumerate(TEST_SETPOINTS):
        log_pid = run_pid(sp)
        log_ag = run_agent(sp, theta)

        m_pid = metrics(log_pid, sp, env_dt)
        m_ag = metrics(log_ag, sp, env_dt)
        rows.append((sp, "PID", m_pid))
        rows.append((sp, "Agente (CEM)", m_ag))

        ax1 = axes[0, j]
        ax1.plot(log_pid["t"], log_pid["T"], label="PID", lw=2)
        ax1.plot(log_ag["t"], log_ag["T"], label="Agente (CEM)", lw=2, ls="--")
        ax1.axhline(sp, color="gray", ls=":", label="Setpoint")
        ax1.set_title(f"Setpoint = {sp:.0f} K")
        ax1.set_ylabel("Temperatura do reator [K]")
        if j == 0:
            ax1.legend(fontsize=8)

        ax2 = axes[1, j]
        ax2.plot(log_pid["t"], log_pid["Tc"], label="PID", lw=2)
        ax2.plot(log_ag["t"], log_ag["Tc"], label="Agente (CEM)", lw=2, ls="--")
        ax2.set_xlabel("Tempo [min]")
        ax2.set_ylabel("Tc (acao de controle) [K]")

    plt.tight_layout()
    plt.savefig("comparacao_pid_vs_agente.png", dpi=150)
    print("Grafico salvo em comparacao_pid_vs_agente.png\n")

    print(f"{'Setpoint':>9} | {'Controlador':>13} | {'ISE':>10} | {'IAE':>10} | {'Overshoot(K)':>13} | {'Acomod.(min)':>13}")
    print("-"*85)
    for sp, name, m in rows:
        print(f"{sp:9.0f} | {name:>13} | {m['ISE']:10.2f} | {m['IAE']:10.2f} | {m['overshoot_K']:13.2f} | {m['settling_time_min']:13.2f}")

if __name__ == "__main__":
    main()
```

```python
%%writefile plant_with_faults.py
"""
Extensao do modelo do CSTR (cstr_model.py) que permite injetar
perturbacoes e falhas ao longo da simulacao, para testar um
supervisor (humano, regras ou LLM) em cima do controle PID/MPC.

Cenarios implementados:
  1. Degrau na temperatura de alimentacao (Tf)          -> disturbio termico
  2. Degrau na concentracao de alimentacao (Caf)         -> disturbio de carga reacional
  3. Setpoint agressivo que satura o atuador (Tc)        -> limite fisico
  4. Falha de sensor: leitura de temperatura com vies     -> instrumentacao com defeito
"""
import numpy as np
from cstr_model import CSTR

class FaultyCSTRSimulation:
    def __init__(self, dt=0.05, total_minutes=25.0, seed=0):
        self.plant = CSTR()
        self.dt = dt
        self.n_steps = int(total_minutes / dt)
        self.rng = np.random.default_rng(seed)
        self.reset()

    def reset(self, Ca0=0.5, T0=350.0):
        self.state = np.array([Ca0, T0], dtype=float)
        self.t = 0
        self.sensor_bias = 0.0
        self.Tc_prev = (self.plant.Tc_min + self.plant.Tc_max) / 2.0
        self.max_dTc_per_step = 5.0  # mesma taxa maxima de variacao usada no CSTREnv
        self.history = {"t": [], "T_true": [], "T_sensor": [], "Ca": [],
                         "Tc": [], "setpoint": [], "Tf": [], "Caf": []}
        return self.state

    def apply_disturbance(self, scenario, k):
        """Ajusta parametros da planta conforme o cenario e o passo atual k."""
        t_min = k * self.dt
        if scenario == "feed_temperature":
            self.plant.Tf = 315.0 if t_min >= 8.0 else 300.0
        elif scenario == "feed_concentration":
            self.plant.Caf = 1.3 if t_min >= 8.0 else 1.0
        elif scenario == "actuator_saturation":
            pass  # o disturbio aqui eh o setpoint agressivo, tratado no run_experiment
        elif scenario == "sensor_fault":
            self.sensor_bias = 10.0 if t_min >= 8.0 else 0.0

    def sensed_temperature(self, T_true):
        return T_true + self.sensor_bias

    def action_to_Tc(self, action_norm):
        """Converte uma acao normalizada em [-1, 1] para Tc fisico,
        respeitando o limite de taxa de variacao do atuador - MESMA
        convencao usada em cstr_env.py, para manter os controladores
        (PID, MPC) compativeis com os ganhos ja ajustados no resto do
        projeto."""
        Tc_center = (self.plant.Tc_min + self.plant.Tc_max) / 2.0
        Tc_half_range = (self.plant.Tc_max - self.plant.Tc_min) / 2.0
        Tc_target = Tc_center + np.clip(action_norm, -1.0, 1.0) * Tc_half_range
        dTc = np.clip(Tc_target - self.Tc_prev, -self.max_dTc_per_step, self.max_dTc_per_step)
        Tc = np.clip(self.Tc_prev + dTc, self.plant.Tc_min, self.plant.Tc_max)
        self.Tc_prev = Tc
        return Tc

    def step(self, Tc):
        self.state = self.plant.step(self.state, Tc, self.dt)
        return self.state

    def log(self, k, setpoint, Tc):
        Ca, T_true = self.state
        T_sensor = self.sensed_temperature(T_true)
        self.history["t"].append((k+1) * self.dt)
        self.history["T_true"].append(T_true)
        self.history["T_sensor"].append(T_sensor)
        self.history["Ca"].append(Ca)
        self.history["Tc"].append(Tc)
        self.history["setpoint"].append(setpoint)
        self.history["Tf"].append(self.plant.Tf)
        self.history["Caf"].append(self.plant.Caf)

    def as_arrays(self):
        return {k: np.array(v) for k, v in self.history.items()}
```

```python
%%writefile supervisor_tools.py
"""
Define as "ferramentas" (tools) que o supervisor - LLM ou baseado em
regras - pode chamar para intervir no controle, e a logica que
aplica essas chamadas ao estado do controlador/planta.

Isso e o "function calling" mencionado na proposta: o supervisor nao
manda um numero direto pra planta a cada milissegundo, ele chama
acoes de mais alto nivel, com validacao e limites de seguranca.
"""
import numpy as np

# Definicao das ferramentas no formato usado pela API da Anthropic
# (JSON Schema) - mesma estrutura serve pra adaptar a OpenAI se quiser.
TOOL_DEFINITIONS = [
    {
        "name": "get_trend",
        "description": "Retorna o historico recente de temperatura (sensor primario) dos ultimos N minutos, para o supervisor analisar a tendencia antes de decidir uma acao.",
        "input_schema": {
            "type": "object",
            "properties": {
                "window_minutes": {"type": "number", "description": "Janela de tempo em minutos a olhar para tras (ex: 3.0)"}
            },
            "required": ["window_minutes"]
        }
    },
    {
        "name": "change_setpoint",
        "description": "Altera o setpoint de temperatura do controlador PID. Use quando o setpoint atual for fisicamente inalcancavel dado os limites do atuador, ou quando for mais seguro operar num ponto diferente temporariamente.",
        "input_schema": {
            "type": "object",
            "properties": {
                "new_setpoint_K": {"type": "number", "description": "Novo setpoint de temperatura em Kelvin"}
            },
            "required": ["new_setpoint_K"]
        }
    },
    {
        "name": "change_pid_gains",
        "description": "Reajusta os ganhos do PID em tempo real (gain scheduling supervisorio). Use quando perceber que o PID esta muito lento, muito agressivo/oscilante, ou com windup excessivo do termo integral.",
        "input_schema": {
            "type": "object",
            "properties": {
                "Kp": {"type": "number"},
                "Ki": {"type": "number"},
                "Kd": {"type": "number"}
            },
            "required": []
        }
    },
    {
        "name": "override_cooling",
        "description": "Sobrepoe manualmente a acao do PID por um curto periodo (emergencia), fixando a temperatura do fluido de resfriamento (Tc) diretamente. Depois de 'duration_minutes', o controle volta automaticamente para o PID. Use com moderacao - e uma acao de emergencia.",
        "input_schema": {
            "type": "object",
            "properties": {
                "Tc_K": {"type": "number", "description": "Temperatura de resfriamento a forcar, em Kelvin"},
                "duration_minutes": {"type": "number", "description": "Por quantos minutos manter o override antes de devolver o controle ao PID"}
            },
            "required": ["Tc_K", "duration_minutes"]
        }
    },
    {
        "name": "flag_sensor_inconsistency",
        "description": "Registra que o supervisor suspeita de uma falha no sensor de temperatura primario (ex: leitura inconsistente com o sensor secundario ou com o comportamento fisico esperado). Nao altera o controle, apenas documenta o diagnostico.",
        "input_schema": {
            "type": "object",
            "properties": {
                "reasoning": {"type": "string", "description": "Breve justificativa do porque suspeita da falha"}
            },
            "required": ["reasoning"]
        }
    },
]

class ControlState:
    """Guarda o estado 'vivo' que o supervisor pode alterar: setpoint,
    ganhos do PID, e um eventual override manual de Tc."""
    def __init__(self, setpoint, Kp, Ki, Kd, Tc_min, Tc_max):
        self.setpoint = setpoint
        self.original_setpoint = setpoint  # para avaliar o supervisor de forma justa depois
        self.Kp, self.Ki, self.Kd = Kp, Ki, Kd
        self.Tc_min, self.Tc_max = Tc_min, Tc_max
        self.override_Tc = None
        self.override_until_step = -1
        self.flags = []
        self.saturation_streak = 0     # quantas consultas seguidas viram saturacao
        self.setpoint_changes_made = 0  # limita quantas vezes o setpoint pode ser recuado

def apply_tool_call(tool_name, tool_input, control_state, current_step, dt, history_arrays):
    """Executa a acao pedida pelo supervisor e retorna uma string de
    resultado (que seria devolvida pra LLM como 'tool_result')."""
    if tool_name == "get_trend":
        window = tool_input.get("window_minutes", 3.0)
        n_back = max(1, int(window / dt))
        t = history_arrays["t"][-n_back:]
        T = history_arrays["T_sensor"][-n_back:]
        pontos = ", ".join(f"{tt:.2f}min:{TT:.1f}K" for tt, TT in zip(t, T))
        return f"Tendencia dos ultimos {window} min -> {pontos}"

    elif tool_name == "change_setpoint":
        new_sp = tool_input["new_setpoint_K"]
        control_state.setpoint = new_sp
        return f"Setpoint alterado para {new_sp:.1f} K"

    elif tool_name == "change_pid_gains":
        if "Kp" in tool_input: control_state.Kp = tool_input["Kp"]
        if "Ki" in tool_input: control_state.Ki = tool_input["Ki"]
        if "Kd" in tool_input: control_state.Kd = tool_input["Kd"]
        return f"Ganhos atualizados: Kp={control_state.Kp}, Ki={control_state.Ki}, Kd={control_state.Kd}"

    elif tool_name == "override_cooling":
        Tc_req = tool_input["Tc_K"]
        duration = tool_input["duration_minutes"]
        if not (control_state.Tc_min <= Tc_req <= control_state.Tc_max):
            return (f"ERRO: Tc={Tc_req:.1f}K fora dos limites fisicos "
                    f"[{control_state.Tc_min}, {control_state.Tc_max}] K. Override rejeitado.")
        control_state.override_Tc = Tc_req
        control_state.override_until_step = current_step + int(duration / dt)
        return f"Override manual aplicado: Tc={Tc_req:.1f}K por {duration:.1f} min"

    elif tool_name == "flag_sensor_inconsistency":
        control_state.flags.append((current_step * dt, tool_input.get("reasoning", "")))
        return "Inconsistencia de sensor registrada no log de diagnostico."

    else:
        return f"ERRO: ferramenta desconhecida '{tool_name}'"

def build_process_summary(history_arrays, control_state, secondary_sensor=None):
    """Monta o texto de resumo do processo que e enviado ao supervisor
    a cada consulta - no mesmo espirito do exemplo da proposta."""
    T_sensor = history_arrays["T_sensor"][-1]
    sp = control_state.setpoint
    Tc = history_arrays["Tc"][-1]
    err = sp - T_sensor

    trend_1min = history_arrays["T_sensor"][-max(1, int(1.0/0.05)):]
    trend_desc = f"{trend_1min[0]:.1f}K -> {trend_1min[-1]:.1f}K" if len(trend_1min) > 1 else "N/D"

    lines = [
        f"TEMPERATURA (sensor primario): {T_sensor:.1f} K",
        f"SETPOINT: {sp:.1f} K",
        f"ERRO: {err:+.1f} K",
        f"TENDENCIA (ultimo minuto): {trend_desc}",
        f"VARIAVEL MANIPULADA Tc: {Tc:.1f} K",
        f"LIMITES Tc: [{control_state.Tc_min:.0f}, {control_state.Tc_max:.0f}] K",
        f"GANHOS PID ATUAIS: Kp={control_state.Kp}, Ki={control_state.Ki}, Kd={control_state.Kd}",
    ]
    if secondary_sensor is not None:
        lines.append(f"SENSOR SECUNDARIO (redundante, independente): {secondary_sensor:.1f} K")
    return "\n".join(lines)
```

```python
%%writefile mock_supervisor.py
"""
Supervisor "de mentirinha" baseado em regras simples, com a MESMA
interface que o supervisor real baseado em LLM (llm_supervisor.py).

Serve para testar toda a "engrenagem" do experimento (cenarios,
ferramentas, metricas, graficos) sem gastar chamadas de API - e como
referencia de comparacao: "supervisor burro por regras" vs "supervisor
LLM" tambem pode virar uma comparacao interessante no TCC.
"""
from supervisor_tools import apply_tool_call

def consult_supervisor(summary_text, control_state, current_step, dt, history_arrays,
                        secondary_sensor=None, verbose=True):
    """Retorna (diagnostico_texto, lista_de_acoes_aplicadas)."""
    T_sensor = history_arrays["T_sensor"][-1]
    sp = control_state.setpoint
    Tc = history_arrays["Tc"][-1]
    err = sp - T_sensor
    actions_log = []

    diagnosis = "Processo dentro do esperado, nenhuma acao necessaria."

    # Regra 1: inconsistencia entre sensores (fault de sensor)
    if secondary_sensor is not None and abs(T_sensor - secondary_sensor) > 5.0:
        diagnosis = (f"Divergencia de {abs(T_sensor-secondary_sensor):.1f} K entre sensor "
                     f"primario e secundario - suspeita de falha de instrumentacao.")
        result = apply_tool_call("flag_sensor_inconsistency",
                                  {"reasoning": diagnosis}, control_state,
                                  current_step, dt, history_arrays)
        actions_log.append(("flag_sensor_inconsistency", {}, result))

    # Regra 2: atuador saturado PERSISTENTEMENTE (nao so uma vez - evita
    # confundir transiente normal de partida com falha real) e erro nao
    # diminui -> recuar setpoint, no maximo 2 vezes, ate um piso minimo
    elif Tc <= control_state.Tc_min + 1.0 and err > 5.0:
        control_state.saturation_streak += 1
        if (control_state.saturation_streak >= 3
                and control_state.setpoint_changes_made < 2):
            diagnosis = ("Atuador de resfriamento saturado ha varias consultas "
                         "seguidas e erro persistente - setpoint provavelmente "
                         "inalcancavel. Recuando setpoint (ajuste limitado).")
            new_sp = max(sp - 10.0, control_state.original_setpoint - 20.0)
            result = apply_tool_call("change_setpoint", {"new_setpoint_K": new_sp},
                                      control_state, current_step, dt, history_arrays)
            actions_log.append(("change_setpoint", {"new_setpoint_K": new_sp}, result))
            control_state.setpoint_changes_made += 1
            control_state.saturation_streak = 0
        else:
            diagnosis = (f"Atuador proximo do limite (streak={control_state.saturation_streak}), "
                         f"monitorando antes de agir.")
    else:
        control_state.saturation_streak = 0

    # Regra 3: erro grande e crescente -> reforcar ganho proporcional temporariamente
    if diagnosis == "Processo dentro do esperado, nenhuma acao necessaria." and abs(err) > 8.0:
        diagnosis = (f"Erro de {err:+.1f} K acima do aceitavel - reforcando ganho "
                     f"proporcional do PID para reagir mais rapido.")
        new_Kp = control_state.Kp * 1.5
        result = apply_tool_call("change_pid_gains", {"Kp": new_Kp},
                                  control_state, current_step, dt, history_arrays)
        actions_log.append(("change_pid_gains", {"Kp": new_Kp}, result))

    if verbose:
        print(f"[t={current_step*dt:5.2f} min] Supervisor (regras): {diagnosis}")
    return diagnosis, actions_log
```

```python
%%writefile run_experiment.py
"""
Orquestra o experimento principal: para cada cenario de falha,
roda "PID sozinho" vs "PID + Supervisor" e compara.

Uso:
    python3 run_experiment.py                 # usa o supervisor mock (regras), gratis e rapido
"""
import argparse
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

from plant_with_faults import FaultyCSTRSimulation
from pid_controller import PID
from supervisor_tools import ControlState, build_process_summary

SCENARIOS = ["feed_temperature", "feed_concentration", "actuator_saturation", "sensor_fault"]
SCENARIO_LABELS = {
    "feed_temperature": "Disturbio: temperatura de alimentacao",
    "feed_concentration": "Disturbio: concentracao de alimentacao",
    "actuator_saturation": "Setpoint agressivo (saturacao do atuador)",
    "sensor_fault": "Falha de sensor (vies de leitura)",
}

BASE_PID_GAINS = dict(Kp=0.0006, Ki=0.2, Kd=0.16)
DEFAULT_SETPOINT = 360.0
AGGRESSIVE_SETPOINT = 420.0  # usado no cenario de saturacao (fisicamente dificil de alcancar)
SUPERVISOR_INTERVAL_MINUTES = 1.0

def run_scenario(scenario, use_supervisor, supervisor_module=None, verbose=False):
    dt = 0.05
    sim = FaultyCSTRSimulation(dt=dt, total_minutes=25.0)
    sim.reset()

    setpoint0 = AGGRESSIVE_SETPOINT if scenario == "actuator_saturation" else DEFAULT_SETPOINT
    cs = ControlState(setpoint=setpoint0, Tc_min=sim.plant.Tc_min, Tc_max=sim.plant.Tc_max,
                       **BASE_PID_GAINS)
    pid = PID(cs.Kp, cs.Ki, cs.Kd, dt=dt, u_min=-1.0, u_max=1.0)

    diagnoses = []
    supervisor_interval_steps = int(SUPERVISOR_INTERVAL_MINUTES / dt)

    for k in range(sim.n_steps):
        sim.apply_disturbance(scenario, k)

        # sincroniza ganhos do PID com o que o supervisor decidiu (se mudou)
        pid.Kp, pid.Ki, pid.Kd = cs.Kp, cs.Ki, cs.Kd

        T_sensor = sim.sensed_temperature(sim.state[1])

        if cs.override_Tc is not None and k < cs.override_until_step:
            Tc = cs.override_Tc
            sim.Tc_prev = Tc  # mantem o rate-limiter consistente apos o override
        else:
            cs.override_Tc = None
            action_norm = pid.compute(cs.setpoint, T_sensor)
            Tc = sim.action_to_Tc(action_norm)

        sim.step(Tc)
        sim.log(k, cs.setpoint, Tc)

        if use_supervisor and k > 0 and k % supervisor_interval_steps == 0:
            secondary = None
            if scenario == "sensor_fault":
                # sensor secundario redundante, sem o vies do primario
                secondary = sim.state[1]
            summary = build_process_summary(sim.as_arrays(), cs, secondary_sensor=secondary)
            diagnosis, actions = supervisor_module.consult_supervisor(
                summary, cs, k, dt, sim.as_arrays(), secondary_sensor=secondary, verbose=verbose)
            diagnoses.append((k*dt, diagnosis, actions))

    return sim.as_arrays(), diagnoses, cs

def metrics_from_arrays(arrays, original_setpoint):
    e_operational = arrays["setpoint"] - arrays["T_true"]
    e_original = original_setpoint - arrays["T_true"]
    dt = arrays["t"][1] - arrays["t"][0]
    return dict(
        ISE_operacional=np.sum(e_operational**2) * dt,
        IAE_operacional=np.sum(np.abs(e_operational)) * dt,
        ISE_vs_setpoint_original=np.sum(e_original**2) * dt,
        IAE_vs_setpoint_original=np.sum(np.abs(e_original)) * dt,
        max_abs_error=np.max(np.abs(e_operational)),
    )

def main(supervisor="mock", verbose=False):
    if supervisor == "llm":
        import llm_supervisor as supervisor_module
    elif supervisor == "ollama":
        import llm_supervisor_ollama as supervisor_module
    else:
        import mock_supervisor as supervisor_module

    fig, axes = plt.subplots(2, len(SCENARIOS), figsize=(5.5*len(SCENARIOS), 7), sharex=True)
    print(f"{'Cenario':>32} | {'Modo':>16} | {'ISE(orig)':>12} | {'IAE(orig)':>10} | {'ErroMax(K)':>10} | {'Setpoint final':>14}")
    print("-"*110)

    for j, scenario in enumerate(SCENARIOS):
        setpoint0 = AGGRESSIVE_SETPOINT if scenario == "actuator_saturation" else DEFAULT_SETPOINT
        arr_pid, _, cs_pid = run_scenario(scenario, use_supervisor=False)
        arr_sup, diagnoses, cs_sup = run_scenario(scenario, use_supervisor=True,
                                              supervisor_module=supervisor_module, verbose=verbose)

        m_pid = metrics_from_arrays(arr_pid, setpoint0)
        m_sup = metrics_from_arrays(arr_sup, setpoint0)
        label = SCENARIO_LABELS[scenario]
        print(f"{label:>32} | {'PID sozinho':>16} | {m_pid['ISE_vs_setpoint_original']:12.1f} | {m_pid['IAE_vs_setpoint_original']:10.1f} | {m_pid['max_abs_error']:10.1f} | {cs_pid.setpoint:14.1f}")
        print(f"{label:>32} | {'PID+Supervisor':>16} | {m_sup['ISE_vs_setpoint_original']:12.1f} | {m_sup['IAE_vs_setpoint_original']:10.1f} | {m_sup['max_abs_error']:10.1f} | {cs_sup.setpoint:14.1f}")

        ax1 = axes[0, j]
        ax1.plot(arr_pid["t"], arr_pid["T_true"], label="PID sozinho", lw=2)
        ax1.plot(arr_sup["t"], arr_sup["T_true"], label="PID + Supervisor", lw=2, ls="--")
        ax1.plot(arr_pid["t"], arr_pid["setpoint"], color="gray", ls=":", lw=1, label="Setpoint")
        for (t_d, diag, actions) in diagnoses:
            if actions:
                ax1.axvline(t_d, color="red", alpha=0.25, lw=1)
        ax1.set_title(SCENARIO_LABELS[scenario], fontsize=9)
        ax1.set_ylabel("Temperatura real [K]")
        if j == 0:
            ax1.legend(fontsize=7)

        ax2 = axes[1, j]
        ax2.plot(arr_pid["t"], arr_pid["Tc"], label="PID sozinho", lw=2)
        ax2.plot(arr_sup["t"], arr_sup["Tc"], label="PID + Supervisor", lw=2, ls="--")
        ax2.set_xlabel("Tempo [min]")
        ax2.set_ylabel("Tc [K]")

    plt.tight_layout()
    plt.savefig("experimento_supervisor.png", dpi=150)
    print("\nGrafico salvo em experimento_supervisor.png")
    print("(linhas vermelhas verticais = momentos em que o supervisor tomou uma acao)")

if __name__ == "__main__":
    main(supervisor="mock", verbose=True)

main(supervisor="mock", verbose=True)
```

```python
# === Chronos no laco de controle: PID vs Agente CEM vs PID + Chronos ===
%pip install -q chronos-forecasting

codigo = r'''"""
Chronos no laco de controle do CSTR: compara 3 controladores no MESMO
cenario de falha (plant_with_faults.py):

1. PID (reativo)          : age sobre a temperatura medida agora
2. Agente CEM             : politica neural treinada (trained_policy.npy)
3. PID + Chronos (predit.): o Chronos preve T nos proximos minutos; o PID
                            age sobre uma mistura da temperatura atual com
                            a temperatura PREVISTA (acao antecipativa)

Tambem mede a qualidade da previsao de T feita pelo Chronos durante a
simulacao (RMSE da previsao vs o que o sensor leu depois).
"""
import time
import numpy as np
import torch
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from chronos import BaseChronosPipeline

from plant_with_faults import FaultyCSTRSimulation
from pid_controller import PID
from policy import MLPPolicy

# ------------------------- configuracao -------------------------
SCENARIO = "feed_temperature"   # "feed_temperature", "sensor_fault", "feed_concentration", "actuator_saturation"
SETPOINT = 360.0
PID_GAINS = dict(Kp=0.0006, Ki=0.2, Kd=0.16)   # mesmos ganhos do resto do projeto
DT = 0.05

CHRONOS_MODEL = "amazon/chronos-bolt-small"
CHRONOS_EVERY = 5      # chama o Chronos a cada 5 passos (0,25 min) e reaproveita a previsao
H_PRED = 20            # horizonte previsto: 20 passos = 1 min
LOOKAHEAD = 10         # PID olha a previsao 10 passos (0,5 min) a frente
ALPHA = 0.5            # peso da previsao: 0 = PID puro, 1 = so a previsao
MIN_CONTEXT = 40       # espera 2 min de historico antes de confiar no Chronos
# -----------------------------------------------------------------

pipe = BaseChronosPipeline.from_pretrained(
    CHRONOS_MODEL,
    device_map="cuda" if torch.cuda.is_available() else "cpu",
    torch_dtype=torch.float32,
)

def chronos_forecast(T_hist):
    ctx = torch.tensor(np.asarray(T_hist[-2048:]), dtype=torch.float32)
    q, _ = pipe.predict_quantiles(ctx, prediction_length=H_PRED, quantile_levels=[0.1, 0.5, 0.9])
    q = q[0].cpu().numpy()                 # [H_PRED, 3]
    return q[:, 1], q[:, 0], q[:, 2]

# normalizacao identica ao CSTREnv (o agente foi treinado com ela)
def agent_obs(T, Ca, sp):
    nT = lambda x: 2 * (x - 300.0) / 150.0 - 1
    return np.array([nT(T), 2 * Ca - 1, nT(sp), (sp - T) / 50.0], dtype=np.float32)

def simulate(controller, theta=None):
    sim = FaultyCSTRSimulation(dt=DT, total_minutes=25.0)
    sim.reset()
    pid = PID(**PID_GAINS, dt=DT)
    policy = MLPPolicy()
    T_meas_hist, forecasts = [], []        # forecasts: (passo_origem, mediana, p10, p90)
    fc = None
    t0 = time.time()
    for k in range(sim.n_steps):
        sim.apply_disturbance(SCENARIO, k)
        Ca, T_true = sim.state
        T_meas = sim.sensed_temperature(T_true)
        T_meas_hist.append(T_meas)

        if controller == "PID":
            a = pid.compute(SETPOINT, T_meas)
        elif controller == "Agente CEM":
            a = float(policy.act(agent_obs(T_meas, Ca, SETPOINT), theta)[0])
        elif controller == "PID + Chronos":
            T_eff = T_meas
            if k >= MIN_CONTEXT:
                if (k - MIN_CONTEXT) % CHRONOS_EVERY == 0:
                    med, p10, p90 = chronos_forecast(T_meas_hist)
                    fc = (k, med, p10, p90)
                    forecasts.append(fc)
                age = k - fc[0]            # passos desde a ultima previsao
                idx = min(age + LOOKAHEAD - 1, H_PRED - 1)
                T_eff = (1 - ALPHA) * T_meas + ALPHA * fc[1][idx]
            a = pid.compute(SETPOINT, T_eff)
        else:
            raise ValueError(controller)

        Tc = sim.action_to_Tc(a)
        sim.step(Tc)
        sim.log(k, SETPOINT, Tc)
    arr = sim.as_arrays()
    arr["T_meas_before_step"] = np.array(T_meas_hist)
    return arr, forecasts, time.time() - t0

def control_metrics(arr, t_fault=8.0):
    e = SETPOINT - arr["T_true"]
    after = arr["t"] >= t_fault
    return {
        "ISE": np.sum(e**2) * DT,
        "IAE": np.sum(np.abs(e)) * DT,
        "IAE pos-falha": np.sum(np.abs(e[after])) * DT,
        "Erro max [K]": np.max(np.abs(e)),
        "Esforco sum|dTc| [K]": np.sum(np.abs(np.diff(arr["Tc"]))),
    }

def forecast_metrics(arr, forecasts):
    """Compara cada previsao com o que o sensor leu nos H_PRED passos seguintes."""
    T = arr["T_meas_before_step"]
    err_la, err_all, inside = [], [], []
    for k0, med, p10, p90 in forecasts:
        real = T[k0 + 1:k0 + 1 + H_PRED]
        n = len(real)
        if n < H_PRED:
            continue
        err_all.append(med[:n] - real)
        err_la.append(med[LOOKAHEAD - 1] - real[LOOKAHEAD - 1])
        inside.append((real >= p10[:n]) & (real <= p90[:n]))
    persist = [T[k0] - T[k0 + LOOKAHEAD] for k0, *_ in forecasts if k0 + LOOKAHEAD < len(T)]
    return {
        f"RMSE Chronos @ {LOOKAHEAD*DT:.2f} min [K]": np.sqrt(np.mean(np.square(err_la))),
        f"RMSE persistencia @ {LOOKAHEAD*DT:.2f} min [K]": np.sqrt(np.mean(np.square(persist))),
        f"RMSE Chronos (horizonte todo, {H_PRED*DT:.1f} min) [K]": np.sqrt(np.mean(np.square(np.concatenate(err_all)))),
        "Cobertura intervalo 10-90%": 100 * np.mean(np.concatenate(inside)),
    }

if __name__ == "__main__":
    theta = np.load("trained_policy.npy")
    runs = {}
    for name in ["PID", "Agente CEM", "PID + Chronos"]:
        arr, fcs, secs = simulate(name, theta)
        runs[name] = (arr, fcs)
        print(f"{name:>14}: simulado em {secs:.1f} s")

    print(f"\nCenario: {SCENARIO} | setpoint {SETPOINT} K | falha em t = 8 min\n")
    names = list(runs)
    mets = {n: control_metrics(runs[n][0]) for n in names}
    keys = list(mets[names[0]])
    print(f"{'Metrica':>22} | " + " | ".join(f"{n:>14}" for n in names))
    print("-" * (25 + 17 * len(names)))
    for key in keys:
        print(f"{key:>22} | " + " | ".join(f"{mets[n][key]:14.2f}" for n in names))

    print("\nQualidade da previsao de T pelo Chronos (durante a simulacao PID + Chronos):")
    for key, v in forecast_metrics(*runs["PID + Chronos"]).items():
        print(f"  {key}: {v:.2f}")

    # --------------------------- graficos ---------------------------
    fig, axes = plt.subplots(3, 1, figsize=(11, 10), sharex=True)
    styles = {"PID": dict(lw=2), "Agente CEM": dict(lw=2, ls="--"), "PID + Chronos": dict(lw=2, ls="-.")}
    for n in names:
        arr = runs[n][0]
        axes[0].plot(arr["t"], arr["T_true"], label=n, **styles[n])
        axes[1].plot(arr["t"], arr["Tc"], label=n, **styles[n])
    axes[0].axhline(SETPOINT, color="gray", ls=":", label="Setpoint")
    axes[0].axvline(8.0, color="red", alpha=0.3, label="Falha")
    axes[0].set_ylabel("T real [K]"); axes[0].legend(fontsize=8)
    axes[0].set_title(f"Controle do CSTR - cenario: {SCENARIO}")
    axes[1].set_ylabel("Tc [K]")

    arr, fcs = runs["PID + Chronos"]
    axes[2].plot(arr["t"], arr["T_meas_before_step"], color="k", lw=1, label="T medida")
    for i, (k0, med, p10, p90) in enumerate(fcs[::8]):
        tt = (k0 + 1 + np.arange(H_PRED)) * DT
        axes[2].plot(tt, med, color="C2", lw=1.5, label="Previsao Chronos" if i == 0 else None)
        axes[2].fill_between(tt, p10, p90, color="C2", alpha=0.2, label="Intervalo 10-90%" if i == 0 else None)
    axes[2].set_ylabel("T [K]"); axes[2].set_xlabel("Tempo [min]")
    axes[2].set_title("Previsoes de T feitas pelo Chronos (1 a cada 2 min mostrada)")
    axes[2].legend(fontsize=8)
    plt.tight_layout(); plt.savefig("chronos_vs_pid_agente.png", dpi=150)
    print("\nGrafico salvo em chronos_vs_pid_agente.png")
'''
open('chronos_control.py', 'w').write(codigo)

!python chronos_control.py
from IPython.display import Image
Image('chronos_vs_pid_agente.png')
```
